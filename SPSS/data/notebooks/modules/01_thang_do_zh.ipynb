# 第01单元 — 数据测量量表与定量数据的本质
第01单元专属 notebook —— 使用 PISA 2025 越南数据集（195所学校）中的真实变量说明四种测量量表。

SPSS/PSPP 对应操作：`Variable View > Measure`、`FREQUENCIES`、`DESCRIPTIVES`。

```python
# 先运行这一格。数据（PISA 2025，越南）直接从 GitHub 获取。
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from scipy import stats
sns.set_theme(style="whitegrid"); plt.rcParams["figure.figsize"]=(7,4); pd.set_option("display.precision",3)

RAW = "https://raw.githubusercontent.com/TatcataiTTN/for-Social-Science/main/SPSS/data/sav/"
truong = pd.read_csv(RAW + "vnm_truong_195_tong_hop.csv")
print("学校表：", truong.shape)
```

#### 📤 实际输出
`(195, 34)` —— 195所学校，与 PISA 2025 越南学校层级数据集一致。

## 1. 定类量表（nominal）— `REGION`
只用于分类，没有内在顺序。

```python
REGION_LABELS={1:"湄公河三角洲",2:"北中部/南中部沿海及西原地区",3:"北部中游及山区",4:"红河三角洲",5:"东南部"}
truong["region_name"] = truong["REGION"].map(REGION_LABELS)
print(truong["region_name"].value_counts())
```

#### 📤 实际输出
5个地区 —— 只是抽样地区的分类标签。可用于计算频数/众数，但不能用于计算平均值（地区3不比地区1「更大」）。

## 2. 定序量表（ordinal）— `SCHSIZE_Q`
国际四分位数（1-4）—— 有顺序，但各级之间的差距不是等量的真实单位。

```python
print(truong["SCHSIZE_Q"].value_counts().sort_index())
print("中位数（因为有顺序所以有效）：", truong["SCHSIZE_Q"].median())
```

## 3. 定距量表（interval）— `EDULEAD`
教育领导力 WLE 指数 —— 单位等距，但「0」只是一个约定的参照点，并不代表「没有领导力」。

```python
print(truong["EDULEAD"].describe().round(3))
```

## 4. 定比量表（ratio）— `W_NRASCHBWT`
学校抽样权重 —— 有真正的零点，比率有意义（「这所学校的权重是那所学校的3倍」是有意义的说法）。

```python
print(truong["W_NRASCHBWT"].describe().round(2))
print(f"权重范围从 {truong['W_NRASCHBWT'].min():.1f} 到 {truong['W_NRASCHBWT'].max():.1f}")
ratio = truong["W_NRASCHBWT"].max() / truong["W_NRASCHBWT"].min()
print(f"权重最高的学校代表的学生数约为权重最低学校的 {ratio:.0f} 倍")
```

#### 📤 实际输出
权重范围从 1.0 到 617.2 —— 这个数字在第02单元中被再次用来解释为何分析 PISA 数据时必须加权。

## 5. 为什么 SPSS/PSPP 的 `Measure` 只有3个选项，而不是4个？
软件把定距和定比合并为一种「Scale」类型，因为大多数统计方法对两者的处理方式相同 —— `EDULEAD`（定距）和 `W_NRASCHBWT`（定比）在 Variable View 中都声明为 Measure=Scale，尽管理论上它们是两种不同的量表。
