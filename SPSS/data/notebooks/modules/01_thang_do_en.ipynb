# Module 01 — Scales of Measurement & the Nature of Quantitative Data
Dedicated notebook for Module 01 — illustrates the 4 scales of measurement using real variables from the PISA 2025 Vietnam dataset (195 schools).

Human-reviewed English translation of the original Vietnamese notebook (code and results are identical, only text is translated).

SPSS/PSPP equivalents: `Variable View > Measure`, `FREQUENCIES`, `DESCRIPTIVES`.

```python
# Run this cell first. Data (PISA 2025, Vietnam) is fetched directly from GitHub.
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from scipy import stats
sns.set_theme(style="whitegrid"); plt.rcParams["figure.figsize"]=(7,4); pd.set_option("display.precision",3)

RAW = "https://raw.githubusercontent.com/TatcataiTTN/for-Social-Science/main/SPSS/data/sav/"
truong = pd.read_csv(RAW + "vnm_truong_195_tong_hop.csv")
print("School table:", truong.shape)
```

#### 📤 Actual output
`(195, 34)` — 195 schools, matching the PISA 2025 Vietnam school-level dataset.

## 1. Nominal scale — `REGION`
Only classifies, no inherent order.

```python
REGION_LABELS={1:"Mekong Delta",2:"North/South Central Coast & Central Highlands",3:"Northern Midlands & Mountains",4:"Red River Delta",5:"Southeast"}
truong["region_name"] = truong["REGION"].map(REGION_LABELS)
print(truong["region_name"].value_counts())
```

#### 📤 Actual output
5 regions — just category labels for the sampling regions. Fine for counting frequencies/mode, NOT meaningful for computing a mean (region 3 is not "greater than" region 1).

## 2. Ordinal scale — `SCHSIZE_Q`
International quartile (1-4) of school size — HAS an order, but the gaps between levels aren't equal real-world units.

```python
print(truong["SCHSIZE_Q"].value_counts().sort_index())
print("Median (valid because it's ordered):", truong["SCHSIZE_Q"].median())
```

## 3. Interval scale — `EDULEAD`
Educational Leadership WLE index — equal-interval units, but "0" is an arbitrary reference point, NOT "no leadership at all".

```python
print(truong["EDULEAD"].describe().round(3))
```

## 4. Ratio scale — `W_NRASCHBWT`
School sampling weight — HAS a true zero, ratios are meaningful ("this school's weight is 3x that school's" is a meaningful statement).

```python
print(truong["W_NRASCHBWT"].describe().round(2))
print(f"Weight ranges from {truong['W_NRASCHBWT'].min():.1f} to {truong['W_NRASCHBWT'].max():.1f}")
ratio = truong["W_NRASCHBWT"].max() / truong["W_NRASCHBWT"].min()
print(f"The highest-weighted school represents about {ratio:.0f}x as many students as the lowest-weighted school")
```

#### 📤 Actual output
Weight ranges from 1.0 to 617.2 → roughly 617x — this exact figure is reused in Module 02 to explain why weighting MUST be applied when analysing PISA data.

## 5. Why do SPSS/PSPP only offer 3 `Measure` options, not 4?
The software merges Interval + Ratio into a single "Scale" type, because most statistical procedures treat them the same way — both `EDULEAD` (interval) and `W_NRASCHBWT` (ratio) are declared as Measure=Scale in Variable View, even though they are theoretically two different scales.
