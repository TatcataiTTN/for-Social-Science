# Module 01 — Thang đo & bản chất dữ liệu định lượng
Notebook RIÊNG cho Module 01 — minh hoạ 4 thang đo dữ liệu bằng chính các biến thật trong bộ dữ liệu PISA 2025 Việt Nam (195 trường).

Tương ứng SPSS/PSPP: `Variable View > Measure`, `FREQUENCIES`, `DESCRIPTIVES`.

```python
# Chạy ô này đầu tiên. Dữ liệu (PISA 2025, Việt Nam) tải trực tiếp từ GitHub.
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from scipy import stats
sns.set_theme(style="whitegrid"); plt.rcParams["figure.figsize"]=(7,4); pd.set_option("display.precision",3)

RAW = "https://raw.githubusercontent.com/TatcataiTTN/for-Social-Science/main/SPSS/data/sav/"
truong = pd.read_csv(RAW + "vnm_truong_195_tong_hop.csv")
print("Bảng trường:", truong.shape)
```

#### 📤 Đầu ra thật
`(195, 34)` — 195 trường, đúng khớp bộ dữ liệu PISA 2025 VN cấp trường.

## 1. Thang định danh (nominal) — `REGION`
Chỉ phân loại, không có thứ tự hơn-kém.

```python
VUNG={1:"ĐB sông Cửu Long",2:"Bắc TB, DH TB & Tây Nguyên",3:"Trung du & MN phía Bắc",4:"ĐB sông Hồng",5:"Đông Nam Bộ"}
truong["vung_ten"] = truong["REGION"].map(VUNG)
print(truong["vung_ten"].value_counts())
```

#### 📤 Đầu ra thật
5 vùng, chỉ là nhãn phân loại khu vực lấy mẫu — hợp lý để tính tần số/mode, KHÔNG hợp lý để tính trung bình (vùng số 3 không "lớn hơn" vùng số 1).

## 2. Thang thứ bậc (ordinal) — `SCHSIZE_Q`
Tứ phân vị quốc tế (1–4) của quy mô trường — CÓ thứ tự, nhưng khoảng cách giữa các mức không đều nhau như đơn vị đo thật.

```python
print(truong["SCHSIZE_Q"].value_counts().sort_index())
print("Median (tính được vì có thứ tự):", truong["SCHSIZE_Q"].median())
```

## 3. Thang khoảng (interval) — `EDULEAD`
Chỉ số lãnh đạo giáo dục (WLE) — có đơn vị đều nhau nhưng "0" là mốc quy ước, KHÔNG có nghĩa "không có lãnh đạo".

```python
print(truong["EDULEAD"].describe().round(3))
```

## 4. Thang tỉ lệ (ratio) — `W_NRASCHBWT`
Trọng số trường — CÓ điểm 0 thật, lấy tỉ số được ("trường này có trọng số gấp 3 trường kia" là câu nói có nghĩa).

```python
print(truong["W_NRASCHBWT"].describe().round(2))
print(f"Trọng số dao động từ {truong['W_NRASCHBWT'].min():.1f} tới {truong['W_NRASCHBWT'].max():.1f}")
ratio = truong["W_NRASCHBWT"].max() / truong["W_NRASCHBWT"].min()
print(f"Trường có trọng số cao nhất đại diện cho số học sinh gấp {ratio:.0f} lần trường thấp nhất")
```

#### 📤 Đầu ra thật
Trọng số từ 1,0 đến 617,2 → gấp ≈617 lần — đây chính là con số dùng lại ở Module 02 để giải thích vì sao PHẢI áp trọng số khi phân tích dữ liệu PISA.

## 5. Vì sao SPSS/PSPP chỉ có 3 lựa chọn `Measure`, không phải 4?
Phần mềm gộp Khoảng + Tỉ lệ thành một loại duy nhất "Scale", vì phần lớn phép thống kê áp dụng như nhau cho cả hai — `EDULEAD` (khoảng) và `W_NRASCHBWT` (tỉ lệ) đều khai báo Measure=Scale trong Variable View, dù về mặt lý thuyết là 2 thang đo khác nhau.
